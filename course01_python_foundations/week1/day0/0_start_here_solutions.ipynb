### Exercise 1

Q. What is a year's coupon income on 1,000,000 face at 4.5 percent?

````python
# face times the coupon in percent, divided by 100
1000000 * 4.5 / 100
````

45,000 a year.
The coupon is a percent, so it is divided by 100.

### Exercise 2

Q. For 100.50, 97.25, and 99.00: the lowest price, and the gap between the highest and the lowest?

````python
# the three new prices, per 100 face, in one list
new_prices = [100.50, 97.25, 99.00]
print("Lowest:", min(new_prices))
print("Gap:", max(new_prices) - min(new_prices))
````

97.25, and a gap of 3.25 points between the highest and the lowest.

### Exercise 3

Q. How many days from 15 January 2026 to 15 July 2026?

````python
from datetime import date

# subtracting two dates gives a length of time; .days reads it in days
days = date(2026, 7, 15) - date(2026, 1, 15)
print(days.days, "days")
````

181 days on the calendar.
Bond day-count conventions, which count some months as 30 days, come in Course 2.

### Exercise 4

Q. How do you draw the three bonds' prices as a bar chart?

````python
import pandas as pd

# the lesson's table: coupon in percent, price per 100 face
bonds = pd.DataFrame({
    "ticker": ["TRNW", "OSMC", "CLDB"],
    "coupon": [4.375, 5.00, 8.25],
    "price": [96.25, 100.00, 104.50],
})
# the chart line from 0.6, with the prices as the bar heights
chart = bonds.plot.bar(x="ticker", y="price", rot=0, title="Price by bond, per 100 face")
````

CLDB, at 104.50, is the tallest bar; only `y=` and the title changed.

### Exercise 6: AI check

Q. Read the assistant's code, run it, and find the bug.

````python
# written for this course in the style of an AI assistant's answer. It contains one bug.
import Pandas as pd

bonds = pd.DataFrame({"ticker": ["TRNW", "OSMC", "CLDB"], "price": [96.25, 100.00, 104.50]})
print(bonds)
````

The code stops at its first line: the library is installed, but its name is `pandas`, all lower case.
Python's names are case-sensitive, so `Pandas` is a different name, and no library has it.
The fix is one letter.

````python
# the library's name is pandas, in lower case
import pandas as pd

bonds = pd.DataFrame({"ticker": ["TRNW", "OSMC", "CLDB"], "price": [96.25, 100.00, 104.50]})
print(bonds)
````

Three rows.
An error message that names the problem is the easy kind of bug: read its last line first.
