### The ^ Trap

Q. What if you type a power the Excel way?

````python
# a power typed the Excel way
5 ^ 2
````

7, not 25, and no error message: a wrong answer that looks like a right one.
In Excel, `^` means "to the power of".
In Python, it does not.
Python uses `^` for a different job, one you will not need on a bond desk, and that job happens to turn 5 and 2 into 7.
All you need to remember: in Python, `^` is never a power.

Common mistake: a power typed the Excel way. You typed 5 ^ 2 and got 7, with no error message. Type 5 ** 2 instead, which gives 25.

**For a power, always type** `**`.

### Order of Operations

Q. Does the order of a sum matter?

````python
# no brackets
2 + 3 * 4
````

````python
# with brackets
(2 + 3) * 4
````

Python multiplies first (2 + 12), but brackets come before that (5 times 4).
Excel agrees: `=2+3*4` gives 14 and `=(2+3)*4` gives 20.

**When in doubt, add brackets.**

## 0.5 What a `.py` File Is

### A File of Code Only

A notebook holds text, code, and results together.
A `.py` file holds only code: plain text, one line of Python under another, and no results.
It is where code goes when you want to keep it and run it again.

**You are about to make one, look at it, run it, and change it.**

### Make the File

Q. How do you make a `.py` file from a notebook?

````python
%%writefile half_coupon.py
# the annual coupon, in percent
coupon = 5.25
# print shows the label, then half the coupon
print("Half the coupon:", coupon / 2)
````

The first line is not Python: it is an instruction to the notebook.
`%%writefile half_coupon.py` says: do not run this cell, save the lines under me as a file called `half_coupon.py`.
`Writing half_coupon.py` is the notebook's answer: the file now exists.
The four lines under the first are the file: two comments and two lines of Python.
`print` shows several things at once when you separate them with commas.

**The file is on the computer now. Go and look at it.**

## 0.6 Libraries

### Built In

A few named commands, called **functions**, come with Python.
`print` shows, `len` counts, `sum` adds, `max` finds the largest, and `min` the smallest.

To use them on several prices at once, first put the prices in a **list**: values inside square brackets, separated by commas.
`[98.50, 101.25, 99.75]` is a list of three prices, like three cells in a row of a spreadsheet.
`prices = [...]` stores the whole list under one name, as `coupon = 5.25` stored one number.

Q. How many prices are there, what is their total, and which are the highest and the lowest?

````python
# three prices, per 100 face, stored together in one list
prices = [98.50, 101.25, 99.75]
print("Count:", len(prices))
print("Total:", sum(prices))
print("Highest:", max(prices))
print("Lowest:", min(prices))
````

Each function takes the whole list and gives back one number.

**Anything more comes in a library, in two steps.**

### Install Once, Import Every Time

Figure 8: Step 1, install once from PyPI with pip. Step 2, import in every session.

A **library** is code someone has already written and tested.
**Step 1, install once.** Some libraries come with Python: the **standard library**.
The rest come from **PyPI**, the Python Package Index, through **pip**, a program that comes with Python.
**Step 2, import every time.** `import` brings an installed library into the notebook, each session, because the kernel starts empty.

**The standard library needs only Step 2.**

### Three Standard Libraries

Q. What is the average price?

````python
import statistics

# library name, a dot, then the function inside it
statistics.mean(prices)
````

About 99.83.

Q. How many days lie between two coupon dates?

````python
from datetime import date

# date(year, month, day): subtract the earlier coupon date from the later one
days = date(2026, 7, 15) - date(2026, 1, 15)
print(days.days, "days")
````

`from datetime import date` takes one piece of a library.
Subtracting dates gives a length of time; `.days` reads it in days.

### Three Bonds in a Table

Q. Can pandas lay out the three bonds like a worksheet?

### Draw the Three Bonds

`x=` names the column along the bottom, `y=` the bar heights.
