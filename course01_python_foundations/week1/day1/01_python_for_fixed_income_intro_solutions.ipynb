### Exercise 1

Q. What is TRNW's annual coupon income, and what did the position cost at the clean price?
Store the answers in **ex1_income** and **ex1_cost**.

```python
# store the coupon (in percent), price (per 100 face), and face held of TRNW
coupon = 4.375
price = 96.25
face = 2000000
# both the coupon and the price are per 100, so divide by 100
ex1_income = face * coupon / 100
ex1_cost = face * price / 100
print(f'Annual coupon income: ${ex1_income:,.2f}')
print(f'Cost: ${ex1_cost:,.2f}')
```

```python
print('Exercise 1 income:', ex1_income, '| expected:', 87500.0)
print('Exercise 1 cost:', ex1_cost, '| expected:', 1925000.0)
```

### Exercise 2

Q. What are the highest coupon, the total face held, and the simple average price of the three bonds?
Store the answers in **ex2_highest_coupon**, **ex2_total_face**, and **ex2_average_price**.

```python
# one list per column, the bonds in the same order in each
ticker_list = ['TRNW', 'OSMC', 'CLDB']
coupon_list = [4.375, 5.0, 8.25]
price_list = [96.25, 100.0, 104.5]
face_list = [2000000, 1250000, 500000]

ex2_highest_coupon = max(coupon_list)
ex2_total_face = sum(face_list)
ex2_average_price = sum(price_list) / len(price_list)
print(f'Highest coupon: {ex2_highest_coupon:.3f} percent')
print(f'Total face held: {ex2_total_face:,}')
print(f'Average price: {ex2_average_price:.2f}')
```

```python
print('Exercise 2 highest coupon:', ex2_highest_coupon, '| expected:', 8.25)
print('Exercise 2 total face:', ex2_total_face, '| expected:', 3750000)
print('Exercise 2 average price:', ex2_average_price, '| expected:', 100.25)
```

### Exercise 3

Q. Can you store the three bonds as a list of dictionaries named **ex3_holdings**?
Then look up 'Rating' in the first one with `get()` and the fallback 'not recorded', and store it in **ex3_rating**.

```python
# one dictionary per bond, in the order of the table
ex3_holdings = [
    {'Ticker': 'TRNW', 'Price': 96.25, 'Face': 2000000},
    {'Ticker': 'OSMC', 'Price': 100.0, 'Face': 1250000},
    {'Ticker': 'CLDB', 'Price': 104.5, 'Face': 500000}
    ]
# position 0 is the first bond; it has no 'Rating' key, so get() hands back the fallback
ex3_rating = ex3_holdings[0].get('Rating', 'not recorded')
print(f'The list holds {len(ex3_holdings)} bonds')
print(f'The rating of the first bond is {ex3_rating}')
```

The dictionaries hold no 'Rating' key, so `get()` handed back the fallback.
With square brackets, `ex3_holdings[0]['Rating']` would have stopped with a `KeyError`.

```python
print('Exercise 3 holdings:', ex3_holdings)
print('expected:', [
    {'Ticker': 'TRNW', 'Price': 96.25, 'Face': 2000000},
    {'Ticker': 'OSMC', 'Price': 100.0, 'Face': 1250000},
    {'Ticker': 'CLDB', 'Price': 104.5, 'Face': 500000}
    ])
print('Exercise 3 rating:', ex3_rating, '| expected:', 'not recorded')
```

### Exercise 4

Q. Can you pull the ticker and maturity out of 'CLDB 8.25 2031' with slicing, and clean ' b ' into 'B'?
Store them in **ex4_ticker**, **ex4_maturity**, and **ex4_rating**.

```python
description = 'CLDB 8.25 2031'
rating_text = ' b '

# the ticker is the first 4 characters
ex4_ticker = description[:4]
# the year is the last 4 characters; int() makes it a whole number
ex4_maturity = int(description[-4:])
# strip() removes the spaces, upper() makes the capital
ex4_rating = rating_text.strip().upper()
print(f'{ex4_ticker} matures in {ex4_maturity} and is rated {ex4_rating}')
```

A negative start counts from the end, so `description[-4:]` is the last four characters, whatever the length of the string.

```python
print('Exercise 4 ticker:', ex4_ticker, '| expected:', 'CLDB')
print('Exercise 4 maturity:', ex4_maturity, '| expected:', 2031)
print('Exercise 4 rating:', ex4_rating, '| expected:', 'B')
```

```python
# no slice: sum() takes the whole list
ex6_total_face = sum(face_list)
print(f'The total face held is {ex6_total_face:,}')
```
