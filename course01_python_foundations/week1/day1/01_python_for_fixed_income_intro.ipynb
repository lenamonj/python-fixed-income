### A Discount Factor and Its Brackets

Q. A bond repays 100 in 5 years. At 5 percent, compounded twice a year, what is that 100 worth today?

Q. What if the brackets are left out?

```python
# ** is a power: 10 half-years at 2.5 percent each
discount_factor = 1 / (1 + 0.05 / 2) ** 10
present_value = 100 * discount_factor
print(f'Discount factor: {discount_factor:.6f}')
print(f'100 in 5 years is worth {present_value:.2f} today')
```

```python
# without brackets: ** first, then /, then +
print(1 / 1 + 0.05 / 2 ** 10)
# with brackets: the sum we meant
print(1 / (1 + 0.05 / 2) ** 10)
# a power comes before a minus sign in front
print(-2 ** 2)
```

Course 2 prices bonds with exactly this line.
1.000049 is nonsense: Python did `2 ** 10` first, then the divisions, then the plus.

**Excel side by side.** Change every `^` to `**` when moving a formula to Python (Day 0's trap).

### Rounding Half to Even

Q. Round 99.125 to two decimals, and 0.5, 1.5, and 2.5 to whole numbers.

Q. Four prices in eighths each sit halfway at 2 decimals. Does rounding change their total?

```python
# round(number, decimals)
print(round(99.125, 2))
# with no decimals given, a whole number
print(round(0.5), round(1.5), round(2.5))
```

```python
# four prices in eighths: add then round, or round then add
added_first = 99.125 + 99.375 + 99.625 + 99.875
rounded_first = round(99.125, 2) + round(99.375, 2) + round(99.625, 2) + round(99.875, 2)
print(f'Added first:   {added_first:.2f}')
print(f'Rounded first: {rounded_first:.2f}')
```

Exactly halfway, Python rounds to the even digit: **banker's rounding**.
Both 398.00: two prices went down and two up, so the errors cancelled.
Eighths are stored exactly, so these four truly sit halfway; most decimals are stored a hair off, which is why `round(2.675, 2)` gives 2.67.

**The rule is older than computers.**

> STORY: "Banker's rounding" is a misleading name, of unclear origin.
> Simon Newcomb's tables of 1883 already said: "write the nearest EVEN number."
> In 1906 Robert Woodward called it "the computer's rule", when a computer was a person, because it errs "as often plus as minus."
> Markets never shared one rule: the EU's 1997 euro rule rounds "exactly half-way" up.
> In January 2008, on the Python developers' list, Tim Peters called it "the rounding method experts want most often", and Python 3.0, released that December, adopted it.

### Clean the Broker's Text

Q. A rating arrives as ' bbb ', with a space on each side. Clean it.

Q. A face amount arrives as the text '1,500,000'. Make it a number.

```python
# the rating as it arrives
rating_text = ' bbb '
# strip() removes spaces at both ends, then upper() makes capitals
rating_clean = rating_text.strip().upper()
print(rating_clean)
# lower() goes the other way
print(rating_clean.lower())
```

```python
# the face amount as it arrives, as text
face_text = '1,500,000'
# replace every comma with nothing, then int() makes a whole number
face_number = int(face_text.replace(',', ''))
print(face_number)
# the originals are unchanged; the brackets show where each one starts and ends
print(f'[{rating_text}] [{face_text}]')
```

Two methods in a row: `upper()` works on what `strip()` handed back.
`int()` cannot read commas, so they go first.
A string never changes in place: a method hands back a new one, so store it.
