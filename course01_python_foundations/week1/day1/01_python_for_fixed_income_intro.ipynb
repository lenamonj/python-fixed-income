Q. What does the `f` add? Print the annual income without it and with it.

```python
# without the f, .2 means 2 digits in total, not 2 decimals
print(f'{annual_income:.2}')
# with the f, as above
print(f'{annual_income:.2f}')
```

## 1.2 How a Notebook Runs

Before going further, it helps to know how a notebook keeps track of things.

Python remembers every variable from the moment its cell is run until the notebook is restarted.

It pays no attention to where a cell sits on the page, only to which cells have been run, and in what order.

Q. `cost` was worked out when `face` was 1,000,000. Since then `face` has been changed to 1,500,000. Has `cost` changed too?

```python
# print the two variables as they stand now
print(f'face is {face:,} and cost is ${cost:,.2f}')
# to get the cost of the larger position, the calculation has to be run again
cost_now = face * price / 100
print(f'Worked out again, the cost is ${cost_now:,.2f}')
```

A variable holds a value, not a formula.

**Excel side by side.** This is the biggest difference from a worksheet.

In Excel, a cell holding `=A1*A2/100` recalculates by itself the moment A1 changes.

A line of code runs when its cell is run, and only then.

One more thing follows from this.

**Running a cell twice does its work twice.** The cell `face = face + 500000` took `face` from 1,000,000 to 1,500,000.

Q. QVNE was bought at 98.50 and is now marked at 99.25. How many points has it gained, and what is that worth on 1,000,000 face?

```python
# the change in price, in points (per 100 face)
points_gained = 99.25 - 98.50
# a point is 1% of face, so divide by 100 to get dollars
value_change = points_gained * 1000000 / 100
print(f'QVNE gained {points_gained:.2f} points, worth ${value_change:,.2f} on 1,000,000 face')
```

0.75 points on 1,000,000 face is $7,500.00.

Prices are quoted per 100 of face, so dividing by 100 turns points into dollars, just as it did for the coupon.

Q. QVNE pays 100 of principal in 5 years. Discounted at 5% a year, compounded twice a year, what is that payment worth today?

```python
# ** raises a number to a power: 10 half-year periods at 2.5% each
discount_factor = 1 / (1 + 0.05 / 2) ** 10
# the present value of 100 paid in 5 years
present_value = 100 * discount_factor
print(f'The discount factor is {discount_factor:.6f}')
print(f'100 paid in 5 years is worth {present_value:.2f} today')
```

`**` is the power operator.

Every price and yield calculation in Course 2 is built from this one line.

**Excel side by side.** In Excel the same discount factor is `=1/(1+0.05/2)^10`, which gives 0.781198.

Excel's power operator is `^`.

Python's is `**`.

What happens if we type the Excel way in Python?

```python
discount_factor = 1 / (1 + 0.05 / 2) ^ 10
```

A `TypeError`.

With decimal numbers, `^` at least stops with an error.

Day 0 showed the worse case: with whole numbers, `5 ^ 2` gave 7 and no warning.

When you move a formula from Excel to Python, change every `^` to `**`.

Q. Does the order of the calculation matter? Write the same discount factor without the brackets.

```python
# without brackets: ** first, then /, then +
print(1 / 1 + 0.05 / 2 ** 10)
# with brackets: the calculation we meant
print(1 / (1 + 0.05 / 2) ** 10)
```

Without brackets the answer is 1.000049, which is nonsense for a discount factor.

Python works out `**` first, then `*`, `/`, `//`, and `%` from left to right, then `+` and `-` from left to right.

Brackets come before all of them.

Excel uses the same order, with one exception: a minus sign in front of a number.

In Excel `=-2^2` gives 4.

In Python `-2 ** 2` gives -4, because the power is worked out before the minus.

Write `(-2) ** 2` when you mean the square of -2.

When in doubt, add brackets.

They cost nothing and the next reader sees what you meant.

Q. A client wants 2,350,000 face of QVNE. The bond trades in round lots of 1,000,000. How many round lots is that, and how much is left over as an odd lot?

```python
order_face = 2350000
lot_size = 1000000
# // divides and rounds down: the number of whole lots
round_lots = order_face // lot_size
# % gives the remainder: what is left after the whole lots
odd_lot = order_face % lot_size
print(f'{round_lots} round lots and an odd lot of {odd_lot:,} face')
```

Two round lots and 350,000 left over.

`//` and `%` are a pair: one gives the whole lots, the other gives what is left.

**Excel side by side.** `=QUOTIENT(2350000,1000000)` gives 2 and `=MOD(2350000,1000000)` gives 350000.

Q. Round a price of 99.125 to two decimals, and round 0.5, 1.5, and 2.5 to whole numbers.

```python
# round(number, decimals)
print(round(99.125, 2))
# with no decimals given, round() gives a whole number
print(round(0.5), round(1.5), round(2.5))
```

Python gives 99.12, and 0, 2, 2.

When a number is exactly halfway, Python rounds to the even digit.

This is called banker's rounding.

**Excel side by side.** Excel's `ROUND` rounds a halfway number away from zero.

`=ROUND(99.125,2)` gives 99.13 and `=ROUND(2.5,0)` gives 3.

On most numbers the two agree.

When a Python result must match a spreadsheet to the cent, check the halfway cases.

Q. Why round to the even digit? Take four prices quoted in eighths. Each one lands exactly halfway at 2 decimals. Compare the total before and after rounding.

```python
# add the four prices first
added_first = 99.125 + 99.375 + 99.625 + 99.875
# round each price to 2 decimals, then add
rounded_first = round(99.125, 2) + round(99.375, 2) + round(99.625, 2) + round(99.875, 2)
print(f'Added first: {added_first:.2f}')
print(f'Rounded first: {rounded_first:.2f}')
```

Both totals are 398.00.

Python sent two prices down and two up, so the rounding errors cancelled.

That is the logic of the even rule: a halfway number goes up about half the time and down about half the time.

**Excel side by side.** Excel sends all four up, so the same total comes to 398.02.

> STORY: "Banker's rounding" is a misleading name. The rule goes back at least to the 1880s, to the people who worked out mathematical tables by hand. Simon Newcomb's tables, copyright 1883, said a number exactly halfway can go either way "so far as mere accuracy is concerned", and added: "A good rule to adopt in such a case is to write the nearest EVEN number." In 1906 Robert Woodward called it "the computer's rule", from a time when a computer was a person, because it will "in the long-run, make this error as often plus as minus."
>
> Financial markets had no single rule. A European Commission paper on the euro described many different market and national conventions, and the EU's own rule for euro conversions, set in 1997, goes the other way: a result "exactly half-way" is "rounded up". Where the name came from is still unclear.
>
> Python settled it in 2008. That January its creator, Guido van Rossum, wrote "I think the consensus is against round-to-even in 3.0". Asked in the same thread whether Python should adopt it, core developer Tim Peters answered yes, wrote that it "is without doubt the rounding method experts want most often", and pointed out that computer hardware already rounded that way. Python 3.0 came out that December, and `round(2.5)` gives 2.

## 1.5 Strings

Much of a desk's data is text: tickers, ratings, identifiers, descriptions.

A string behaves like a list of characters, so indexing, slicing, and `len()` all work on it.

Every bond has a 9-character identifier called a CUSIP.

The first 6 characters identify the issuer, the next 2 the issue, and the last is a check digit.

The CUSIP below is invented.

Q. Store the CUSIP of QVNE and pull out its three parts.

```python
# an invented CUSIP for QVNE
cusip = '99012QAB6'
# len() counts the characters
print(len(cusip))
# the first 6 characters: start at the beginning and stop before index 6
print(cusip[:6])
# the next 2 characters: index 6 and index 7
print(cusip[6:8])
# the last character
print(cusip[-1])
```

9 characters, split into 99012Q, AB, and 6.

A slice with nothing before the colon starts at the beginning.

With nothing after the colon, it runs to the end.

**Excel side by side.** With the CUSIP in A1, `=LEN(A1)` returns 9, `=LEFT(A1,6)` returns 99012Q, `=MID(A1,7,2)` returns AB, and `=RIGHT(A1,1)` returns 6.

`MID` counts from 1, so its 7 is Python's index 6.

Text typed by people arrives untidy.

A string has methods that hand back a cleaned-up copy.

Q. A rating arrives as ' bbb ', in lower case with a space on each side. A face amount arrives as the text '1,500,000'. Clean both.

```python
rating_text = ' bbb '
# strip() removes the spaces at both ends, and upper() changes the letters to capitals
rating_clean = rating_text.strip().upper()
print(rating_clean)
# lower() goes the other way
print(rating_clean.lower())

face_text = '1,500,000'
# replace() swaps one piece of text for another. Here every comma is replaced with nothing
face_number = int(face_text.replace(',', ''))
print(face_number)
# the two original strings are unchanged. The square brackets show where each one starts and ends
print(f'[{rating_text}] [{face_text}]')
```

`' bbb '` became `'BBB'`, and the text `'1,500,000'` became the number 1500000.

`int()` cannot read a number with commas in it, so the commas have to go first.

A string cannot be changed in place.

Each method hands back a new string and leaves the original alone, so the result has to be stored in a variable.

**Excel side by side.**

One difference: `TRIM` also squeezes a run of spaces inside the text down to a single space.

`strip()` touches only the two ends.

Q. A description arrives as the single string 'QVNE 5.25 2033'. Split it into its parts. Then check whether the issuer name 'Quorvane Energy' contains the word 'Energy'.

```python
description = 'QVNE 5.25 2033'
# split() cuts a string at the spaces and returns a list of the pieces
parts = description.split()
print(parts)
# every piece is still a string, so convert the coupon before using it as a number
print(float(parts[1]))

issuer = 'Quorvane Energy'
# 'in' asks whether one string appears inside another. The answer is True or False
print('Energy' in issuer)
# upper and lower case must match
print('energy' in issuer)
```

`split()` returned a list of three strings.

With something in the brackets, such as `split(',')`, it cuts at that text in place of the spaces.

`in` found 'Energy' and did not find 'energy'.

To ignore case, compare in lower case: `'energy' in issuer.lower()` is True.

**Excel side by side.** `=ISNUMBER(FIND("Energy",A1))` is TRUE when the text is found and FALSE when it is not.

`FIND` is case-sensitive, like `in`, and `SEARCH` is the version that ignores case.

To split text, Excel has Text to Columns on the Data tab, and Microsoft 365 has the function `=TEXTSPLIT(A1," ")`.
