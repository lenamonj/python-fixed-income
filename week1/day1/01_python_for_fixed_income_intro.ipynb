## 1.3 Conditional Statements

A bond priced above 100 trades at a premium, and a bond priced below 100 trades at a discount. The desk wants each bond labelled.

Q. Write code that prints whether QVNE trades at a discount.

Sometimes a block of code should run only when a condition is true. That is what a conditional statement does.

The general form of an if-else statement is:

```
if condition:
    code to run when the condition is true
else:
    code to run when it is false
```

The indentation is part of the language. Python uses it to know which lines belong to the `if` and which belong to the `else`, so be careful with it.

````python
# store the price of QVNE
price = 98.5

# if-else statement
if price < 100:
    print('QVNE trades at a discount')
else:
    print('QVNE does not trade at a discount')
````

When there are more than two outcomes, `elif` (short for else if) adds more conditions:

```
if condition:
    code
elif another condition:
    code
else:
    code
```

Q. Label DNMR as trading at a premium, at a discount, or at par.

````python
# store the price of DNMR
price = 103.25

if price > 100:
    print('DNMR trades at a premium')
elif price < 100:
    print('DNMR trades at a discount')
else:
    print('DNMR trades at par')
````

* Python checks the conditions from top to bottom and runs the first block that is true. The rest are skipped.
* `==` tests whether two values are equal. A single `=` stores a value.

**Excel side by side.** With the price in A1, the same three-way label in Excel is a nested IF: `=IF(A1>100,"Premium",IF(A1<100,"Discount","Par"))`. Python's `if`, `elif`, `else` does the same job, laid out one condition per line.

## 1.4 Looping Statements

The desk wants to know what its QVNE position of 1,500,000 face would be worth at prices of 96, 97, 98, 99, and 100. Writing the same calculation five times is slow and error-prone. A loop repeats it for us.

Q. Print the value of the QVNE position at each of those prices.

First, a look at the `range()` function, which produces a sequence of whole numbers.

````python
print(range(5))
````

* On its own a range does not show its numbers. Wrapping it in `list()` does.

````python
print(list(range(5)))
````

* `range(5)` starts at 0 and stops before 5.

````python
print(list(range(96, 101)))
````

* With two numbers, it starts at the first and stops before the second.

````python
print(list(range(90, 101, 5)))
````

* A third number sets the step.

A **for loop** runs a block of code once for each item in a sequence:

```
for item in sequence:
    code to repeat
```

````python
# store the face held of QVNE
face = 1500000
# start the for loop
for test_price in range(96, 101):
    # value of the position at this price
    value = face * test_price / 100
    # print the result
    print('At a price of ', test_price, ' the position is worth $', value, sep='')
````

* The indented lines ran five times. Each time, `test_price` held the next number from the range.

A loop can also walk through a list directly.

Q. Print each ticker in the portfolio.

````python
for ticker in ticker_list:
    print(ticker)
````

Q. Print the cost of each position: face times price divided by 100.

* `zip()` pairs up the items of several lists, so that a loop can step through them together.

````python
for ticker, face, price in zip(ticker_list, face_list, price_list):
    # cost of this position
    cost = face * price / 100
    print(ticker, 'cost $', cost)
````

**Excel side by side.** In Excel you would type `=B2*C2/100` in one row and fill it down the column. The loop is the fill-down: one formula, applied to every row.

Q. What did the whole portfolio cost?

````python
# start a running total at zero
total_cost = 0
for face, price in zip(face_list, price_list):
    # add the cost of this position to the running total
    total_cost = total_cost + face * price / 100

print('The portfolio cost $', total_cost, sep='')
````

* The running total is a pattern you will use constantly: start at zero, then add to it inside the loop.

A **while loop** repeats as long as a condition stays true:

```
while condition:
    code to repeat
```

Q. The QVNE position pays a coupon of 39,375 dollars every six months. How many payments does it take to collect at least 150,000 dollars?

````python
# store the size of one coupon payment
payment = 39375
# start with nothing collected and no payments counted
collected = 0
payments = 0
# keep going while the total is below the target
while collected < 150000:
    # receive one more payment
    collected = collected + payment
    # count it
    payments = payments + 1

print('It takes', payments, 'payments to collect $', collected)
````

* A for loop runs a known number of times. A while loop runs until something happens, so we do not need to know the count in advance.
* The condition must eventually become false. If it never does, the loop never stops.

## 1.5 List Comprehensions

The desk wants a new list holding the annual coupon income of each position.

We already have **face_list** and **coupon_list**.

One way is to start with an empty list, loop over the positions, and add each result to the list.

````python
income_list = []

for face, coupon in zip(face_list, coupon_list):
    income = face * coupon / 100
    income_list.append(income)

print(income_list)
````

A **list comprehension** produces the same list in one line.

* It builds a new list from an existing sequence.
* It is shorter than a loop, and once you are used to it, easier to read.

````python
income_list = [face * coupon / 100 for face, coupon in zip(face_list, coupon_list)]
print(income_list)
````

The comprehension has three parts:

1. `[ ]`: the square brackets say we are building a list
2. `face * coupon / 100`: the calculation to perform for each item
3. `for face, coupon in zip(face_list, coupon_list)`: where the items come from

Q. Create a list that labels each bond as 'Premium' or 'Discount' based on its price.

````python
# an if-else inside the comprehension chooses the label
label_list = ['Premium' if price > 100 else 'Discount' for price in price_list]
print(label_list)
````

To see how that line is built, here is the same logic written as a loop.

````python
label_list = []

for price in price_list:
    if price > 100:
        label_list.append('Premium')
    else:
        label_list.append('Discount')

print(label_list)
````

Q. Create a list of only the tickers whose bonds trade at a discount.

````python
# an if at the end keeps only the items that pass the test
discount_tickers = [ticker for ticker, price in zip(ticker_list, price_list) if price < 100]
print(discount_tickers)
````

## 1.6 Functions in Python

**What is a function?**

A function is a named block of code that does one job.

We have already used several: `print`, `len`, `min`, `max`, and `sum` are functions that come with Python.

We can also write our own. Once a calculation lives in a function, we can use it again and again without retyping it.

Q. Write a function that prints a description of the QVNE bond.

````python
# define the function
def describe_qvne():
    """
    Print a description
    of the QVNE bond.
    """
    # store the attributes of the bond
    ticker = 'QVNE'
    coupon = 5.25
    maturity = 2033
    price = 98.5
    # print the description
    print(ticker, ' ', coupon, '% of ', maturity, ' is priced at ', price, sep='')
````

**How is a function used?**

Defining a function does not run it. Nothing printed above. To run it, we **call** it by writing its name followed by round brackets. A function must be defined before it is called.

````python
# call the function
describe_qvne()
````

**How do we make a function work for any bond?**

That function only knows about QVNE. To make it general, we give it **arguments**: inputs that are filled in when the function is called.

Q. Write a function that prints a description of any bond.

````python
# define the function
def describe_bond(ticker, coupon, maturity, price):
    """
    Print a description
    of any bond.
    """
    print(ticker, ' ', coupon, '% of ', maturity, ' is priced at ', price, sep='')
````

````python
# call the function for DNMR
describe_bond('DNMR', 6.125, 2034, 103.25)
````

````python
# call the function for PLWK
describe_bond('PLWK', 4.75, 2031, 95.0)
````

The function expects four arguments and we gave it four, so it runs.

**What is a return statement?**

Printing shows a result on the screen, but the result cannot be used afterwards. A `return` statement hands the result back, so it can be stored in a variable or used in another calculation.

Current yield is the annual coupon divided by the price.

Q. Write a function that takes a coupon and a price and returns the current yield in percent.

````python
# define the function
def current_yield(coupon, price):
    """
    Take the coupon (in percent) and the price (per 100 face)
    and return the current yield in percent.
    """
    # annual coupon divided by price, as a percent
    result = coupon / price * 100
    # hand the result back
    return result
````

````python
# call the function for QVNE and store the result
qvne_yield = current_yield(5.25, 98.5)
qvne_yield
````

* QVNE has a current yield of about 5.33 percent. It is above the 5.25 coupon because the bond costs less than 100.

Because the function returns a value, we can use it inside a loop.

Q. Print the current yield of every bond in the portfolio, rounded to 2 decimals.

````python
for ticker, coupon, price in zip(ticker_list, coupon_list, price_list):
    # round() trims the result to 2 decimals
    print(ticker, 'has a current yield of', round(current_yield(coupon, price), 2), 'percent')
````

There is a second, shorter way to write a small function: the **lambda** function.

## 1.7 \*args and \**kwargs

The desk wants one number: the total face held across its four positions.

We can write a function with four arguments, one for each position.

````python
def total_face(face1, face2, face3, face4):
    """
    Take the face held in four positions
    and return the total.
    """
    total = face1 + face2 + face3 + face4
    return total
````

````python
print('Total face held:', total_face(1500000, 500000, 750000, 250000))
````

This works for exactly four positions. A real portfolio holds hundreds, and the number changes every day. Rewriting the function each time is not practical.

* We want a function that accepts any number of values.

**Putting a \* in front of an argument name lets a function accept any number of values.** By convention the name is `args`.

````python
def total_face(*args):
    """
    Take the face held in any number of positions
    and return the total.
    """
    total = 0
    # add up every value that was passed in
    for arg in args:
        total = total + arg
    return total
````

Let's total the four positions.

````python
print('Total face held:', total_face(1500000, 500000, 750000, 250000))
````

Now the desk adds two more positions.

````python
print('Total face held:', total_face(1500000, 500000, 750000, 250000, 300000, 200000))
````

The name does not have to be `args`. Any name works, and a meaningful one reads better.

````python
def total_face(*positions):
    """
    Take the face held in any number of positions
    and return the total.
    """
    total = 0
    for position in positions:
        total = total + position
    return total
````

````python
print('Total face held:', total_face(1500000, 500000, 750000, 250000, 300000, 200000))
````

So far, every argument has been matched by its position: the first value goes to the first name, and so on. These are **positional arguments**.

The other kind is the **keyword argument**, which has a name and a default value. If we do not supply it, the default is used.

The desk wants the total coupon income from any number of positions. Usually it wants a full year, but sometimes only part of a year.

Let's write a function with a keyword argument for the number of months.

````python
def total_income(*incomes, months=12):
    """
    Take the annual coupon income of any number of positions
    and the number of months,
    and return the total income for that period.
    """
    total = 0
    for income in incomes:
        total = total + income
    # scale the annual total to the period asked for
    return total * months / 12
````

````python
print('Income for a full year:', total_income(78750.0, 30625.0, 35625.0, 18750.0))
````

We did not say how many months, so the function used the default of 12. For six months, we set the keyword argument.

````python
print('Income for six months:', total_income(78750.0, 30625.0, 35625.0, 18750.0, months=6))
````

**Note**: when calling a function, keyword arguments go after the positional arguments.

````python
print('Income for six months:', total_income(months=6, 78750.0, 30625.0, 35625.0, 18750.0))
````

* A `SyntaxError`. Python stops before running anything, because a positional argument came after a keyword argument.

Just as \*args accepts any number of positional arguments, \**kwargs accepts any number of keyword arguments. They arrive inside the function as a dictionary.

On top of coupons, the desk has other cash coming in and going out: a bond maturing, and custody fees to pay. The list of items differs from month to month.

Let's write a function that returns the net cash for the period.

````python
def net_cash(*incomes, months=12, **other_cash):
    """
    Take the annual coupon income of any number of positions,
    the number of months, and any other named cash amounts,
    and return the net cash for the period.
    """
    total = 0
    for income in incomes:
        total = total + income
    total = total * months / 12
    # add every other cash amount, whatever it is called
    for amount in other_cash.values():
        total = total + amount
    return total
````

````python
# other cash for the period: money in is positive, money out is negative
other = {'maturing_bond': 250000, 'custody_fees': -1500}
print('Net cash for six months:', net_cash(78750.0, 30625.0, 35625.0, 18750.0, months=6, **other))
````

* The two stars in the call unpack the dictionary into keyword arguments. Inside the function they are collected again into `other_cash`.

A function can return more than one value.

Q. Write a function that returns the total face, the total cost, and the total annual coupon income of the portfolio.

````python
def portfolio_summary(faces, prices, coupons):
    """
    Take lists of face amounts, prices, and coupons
    and return the total face, total cost, and total annual coupon income.
    """
    total_face = 0
    total_cost = 0
    total_income = 0
    for face, price, coupon in zip(faces, prices, coupons):
        total_face = total_face + face
        total_cost = total_cost + face * price / 100
        total_income = total_income + face * coupon / 100
    # return three values at once
    return total_face, total_cost, total_income
````
