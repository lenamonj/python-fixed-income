# Problem Statement

## Business Context

A credit team has worked with a public file of 30,000 card accounts since Day 6.
The logistic model gives one equation, and a credit committee asks for the logic as plain rules, such as "two or more months late in September, and a September bill above 577 NT dollars".
A decision tree gives exactly that: a stack of yes or no questions fitted to the accounts.
The accounts are real, from Taiwan in 2005, so they describe consumer credit there and then, not US credit today.
The four person columns are never features, because a model that used them would sort people by them.

## Objectives

Fit a tree and read it, see why a tree left alone memorises its accounts, and rein it in three ways: class weights, stopping it early, and cutting it back.
Choose each setting by cross-validation inside the training rows, compare the trees on validation rows, and say which columns a tree leans on.

## Data Dictionary

`credit_card_default.csv`, one row per account:

1. `LIMIT_BAL`: credit limit, NT dollars
2. `PAY_0`, `PAY_2` to `PAY_6`: repayment status, September back to April 2005, where -1 is paid on time, 1 to 9 is months late, and -2 and 0 are undocumented (used as numbers, a modelling choice, as on Day 6)
3. `BILL_AMT1` to `BILL_AMT6`: the bill, NT dollars, September back to April
4. `PAY_AMT1` to `PAY_AMT6`: the amount paid, NT dollars, September back to April

# Model Building

## The Words of a Tree

A tree two questions deep: the root asks if one column is at most t, yes goes left to a second question, no goes right to a leaf, and the second question ends in two leaves, with depth 0, 1 and 2 marked

**Split**: a question of the form "is this column at most t?".

Yes goes left, no goes right.

**Root**: the first split, asked of every account.

**Leaf**: where a path ends.

A tree's output for an account is its leaf's share of defaults among the training accounts.

**Depth**: the number of questions on the longest path from the root to a leaf.

## A Tree With One Question

```python
# a tree allowed one question (a stump), fitted to the training accounts
stump = DecisionTreeClassifier(max_depth=1, random_state=SEED)
stump.fit(X_train, y_train)

# the tree as text: each question, then [paid, defaulted] training accounts in each leaf
print(export_text(stump, feature_names=FEATURES, show_weights=True))
```

Of 19 columns and every threshold, the tree chose `PAY_0 <= 1.5`: the jump between codes 1 and 2 in the default-rate bars.

Left: 16,115 accounts, 16.5 percent defaulted.

Right: 1,885 accounts, 70.1 percent.

1.5 sits halfway between codes 1 and 2.

`random_state=SEED` settles ties between equally good questions the same way on every run.

## Scoring a Tree on the Validation Rows

Day 8's scoring function, repeated: ROC AUC, then Day 7's four measures at a threshold.

```python
# flag at or above the threshold, then ROC AUC and four measures in one row
def model_performance_classification(model, predictors, target, threshold=0.5):
    flags = (model.predict_proba(predictors)[:, 1] >= threshold).astype(int)
    return pd.DataFrame({
        "ROC AUC": [roc_auc_score(target, model.predict_proba(predictors)[:, 1])],
        "Accuracy": [accuracy_score(target, flags)],
        "Recall": [recall_score(target, flags, zero_division=0)],
        "Precision": [precision_score(target, flags, zero_division=0)],
        "F1": [f1_score(target, flags, zero_division=0)],
    })
```

```python
# the one-question tree on accounts it has never seen
model_performance_classification(stump, X_valid, y_valid)
```

The stump gives every account one of two outputs, 0.165 or 0.701.

That alone scores a ROC AUC of 0.6361, well above a coin toss's 0.5.

At 0.5 it flags the right side: 68 percent of its flags are defaults, and it catches 31 percent of them.

**One question already ranks unseen accounts. Next, how the tree picked it.**

## How the Tree Chose: Gini Impurity

The **Gini impurity** of a group whose default share is p is 1 - p^2 - (1 - p)^2.
It is 0 for a group of one outcome and 0.5 for a half and half group.
A split is scored by the two sides' Gini, weighted by their accounts, and the tree takes the split with the largest drop from the parent's Gini.

```python
# Gini impurity of a group whose default share is p
def gini(p):
    return 1 - p**2 - (1 - p)**2

# the root, and the two sides of the question PAY_0 <= 1.5
left = X_train["PAY_0"] <= 1.5
root_gini = gini(y_train.mean())
left_gini, right_gini = gini(y_train[left].mean()), gini(y_train[~left].mean())
weighted = left.mean() * left_gini + (1 - left.mean()) * right_gini
print("Root:", round(root_gini, 4), " Left:", round(left_gini, 4), " Right:", round(right_gini, 4))
print("Weighted:", round(weighted, 4), " Decrease:", round(root_gini - weighted, 4))
```

```python
# the tree's own impurity for the root, the left leaf and the right leaf
stump.tree_.impurity.round(4)
```

The tree's numbers match ours: Gini falls from 0.3446 to 0.2906 on the weighted average, a decrease of 0.0539.

### The Decrease as a Picture

```python
# the Gini curve, the two sides joined by a straight chord, and the parent between them
p = np.linspace(0, 1, 101)
p_root, p_left, p_right = y_train.mean(), y_train[left].mean(), y_train[~left].mean()
plt.plot(p, gini(p), color="gray", label="Gini impurity")
plt.plot([p_left, p_right], [left_gini, right_gini], "o-", label="the two sides, joined")
plt.plot([p_root, p_root], [weighted, root_gini], color="red", linewidth=4, label="the decrease, 0.0539")
plt.scatter([p_root], [root_gini], color="black", zorder=3, label="the root, 0.3446")
plt.text(p_left, left_gini - 0.03, "left side", ha="center")
plt.text(p_right, right_gini - 0.03, "right side", ha="center")
plt.xlabel("Default share in the group")
plt.ylabel("Gini impurity")
plt.title("A Split Lowers Impurity by the Gap Between the Curve and the Chord")
plt.legend(loc="lower center")
plt.show()
```

The weighted Gini of the two sides always sits on the chord, below the curve, at the parent's default share.

The further apart the two sides' shares, the bigger the gap.

### Every Threshold on PAY_0

```python
# the decrease for a question "column <= t" on the training accounts
def gini_decrease(column, t):
    left = X_train[column] <= t
    weighted = left.mean() * gini(y_train[left].mean()) + (1 - left.mean()) * gini(y_train[~left].mean())
    return gini(y_train.mean()) - weighted

# every threshold halfway between two codes, the chosen one highlighted
thresholds = np.arange(-1.5, 8, 1.0)
decreases = [gini_decrease("PAY_0", t) for t in thresholds]
plt.bar(thresholds.astype(str), decreases, color=["C1" if t == 1.5 else "C0" for t in thresholds])
plt.xlabel("Threshold t in the question PAY_0 <= t")
plt.ylabel("Gini decrease")
plt.title("1.5 Gives the Largest Decrease; 0.5 Comes Close, 2.5 Falls Away")
plt.show()
```

1.5 gives 0.0539, 0.5 gives 0.0487, and 2.5 only 0.0080.

The tree runs this search over all 19 columns and keeps the best question.

`criterion="entropy"` is the other common impurity measure, and on these accounts it picks the same root.

**Fitting a tree is this search, repeated inside each side. Nothing tells it when to stop.**

## Where Overfitting Starts

Q. How deep can a tree grow before the validation accounts stop gaining?

```python
# training and validation ROC AUC for trees allowed 1 to 20 questions deep
depths = list(range(1, 21))
train_auc, valid_auc = [], []
for depth in depths:
    depth_tree = DecisionTreeClassifier(max_depth=depth, random_state=SEED).fit(X_train, y_train)
    train_auc.append(roc_auc_score(y_train, depth_tree.predict_proba(X_train)[:, 1]))
    valid_auc.append(roc_auc_score(y_valid, depth_tree.predict_proba(X_valid)[:, 1]))
```

```python
# the two curves, the gap between them shaded, the best validation depth ringed
best = valid_auc.index(max(valid_auc))
plt.plot(depths, train_auc, marker="o", label="training")
plt.plot(depths, valid_auc, marker="o", label="validation")
plt.fill_between(depths, valid_auc, train_auc, color="gray", alpha=0.2, label="the gap")
# facecolor="none" leaves the marker empty, so only its red edge shows: a ring
plt.scatter(depths[best], valid_auc[best], s=250, facecolor="none", edgecolor="red", linewidth=2, zorder=3)
plt.xticks(depths)
plt.xlabel("max_depth (questions on the longest path)")
plt.ylabel("ROC AUC")
plt.title("Validation ROC AUC Peaks at Depth 5, Then Falls as the Gap Opens")
plt.legend()
plt.show()
```

Day 5's **validation curve** again, now for a tree's depth.

Validation ROC AUC peaks at depth 5 (0.7538), and the gap opens after it.

At depth 20 it is 0.9862 on training accounts and 0.5950 on validation accounts.

Q. What map does such a tree draw, with the September status code across and the September bill up the side? Sketch it for a tree two questions deep and for one twenty deep.

```python
# a fine grid over the map, and 400 validation accounts to draw on top, jittered so the codes do not overlap
two = ["PAY_0", "BILL_AMT1"]
codes, bills = np.meshgrid(np.linspace(-2.5, 4.5, 300), np.linspace(-20000, 250000, 300))
grid = pd.DataFrame({"PAY_0": codes.ravel(), "BILL_AMT1": bills.ravel()})
dots = X_valid[two].sample(400, random_state=SEED)
dot_codes = dots["PAY_0"] + np.random.default_rng(SEED).uniform(-0.25, 0.25, len(dots))
dot_colours = y_valid[dots.index].map({0: "white", 1: "black"})
```

```python
# fit a tree of one depth on the two columns, colour the grid by its leaves, and put the accounts on top
def tree_map(max_depth, ax):
    model = DecisionTreeClassifier(max_depth=max_depth, random_state=SEED).fit(X_train[two], y_train)
    share = model.predict_proba(grid)[:, 1].reshape(codes.shape)
    shading = ax.contourf(codes, bills / 1000, share, levels=np.linspace(0, 1, 11), cmap="RdBu_r")
    ax.scatter(dot_codes, dots["BILL_AMT1"] / 1000, c=dot_colours, edgecolor="gray", s=18)
    train_score = roc_auc_score(y_train, model.predict_proba(X_train[two])[:, 1])
    valid_score = roc_auc_score(y_valid, model.predict_proba(X_valid[two])[:, 1])
    ax.set_title(f"Depth {max_depth}: {model.get_n_leaves()} leaves\nROC AUC {train_score:.4f} training, {valid_score:.4f} validation")
    ax.set_xlim(-2.5, 4.5)
    ax.set_ylim(-20, 250)
    ax.set_xlabel("September status code (PAY_0)")
    ax.set_ylabel("September bill (thousand NT dollars)")
    return shading
```

# Model Performance Improvement

## Class Weights

An unweighted leaf says defaulted only where defaults are the majority, and 22 percent of the accounts defaulted.
`class_weight="balanced"` counts each account in inverse proportion to its class's size, so the two outcomes weigh the same in total.

```python
# the weight of one paid account and of one default when both outcomes weigh the same in total
n = len(y_train)
print("Paid:", round(n / (2 * (y_train == 0).sum()), 4), " Defaulted:", round(n / (2 * y_train.sum()), 4))
```

## Pre-Pruning: Stop the Tree Early

**Pre-pruning** sets limits before the fit: `max_depth` caps the questions, and `min_samples_leaf` refuses any split that would leave a leaf with fewer accounts.
A setting chosen before the fit is a **hyperparameter**.
Choosing it on the validation rows would spend them, so we choose inside the training rows by cross-validation.

The 30,000 accounts cut into 18,000 training, 6,000 validation and 6,000 test; the training rows cut into five folds, each round fitting on four folds and scoring the fifth, then the five scores averaged

`StratifiedKFold` cuts the training rows into five folds with the same default share.

`GridSearchCV` tries every pair of settings in a grid, fits on four folds, scores the fifth, and averages the five scores.

It then refits the best pair on all the training rows, as `best_estimator_`.

```python
# five folds of the training accounts, shuffled with the seed
folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

# every pair of settings, each scored by ROC AUC averaged over the five folds
grid_pre = {"max_depth": [4, 6, 8, 10, 12], "min_samples_leaf": [25, 50, 100, 200, 400]}
pre_search = GridSearchCV(DecisionTreeClassifier(class_weight="balanced", random_state=SEED), grid_pre,
                          cv=folds, scoring="roc_auc")
pre_search.fit(X_train, y_train)
print(pre_search.best_params_, round(pre_search.best_score_, 4))
```

```python
# the 25 cross-validated scores, depth down the side and leaf size across, the best outlined
scores = pd.DataFrame(pre_search.cv_results_).pivot(index="param_max_depth", columns="param_min_samples_leaf",
                                                     values="mean_test_score")
ax = sns.heatmap(scores, annot=True, fmt=".4f", cmap="Blues", cbar_kws={"label": "Cross-validated ROC AUC"})
# plt.Rectangle((column, row), 1, 1) is a box one cell wide, and add_patch draws it on the chart
ax.add_patch(plt.Rectangle((3, 2), 1, 1, fill=False, edgecolor="red", linewidth=3))
ax.set_xlabel("min_samples_leaf (accounts)")
ax.set_ylabel("max_depth (questions)")
ax.set_title("Deep Trees Need Big Leaves: Depth 8 With 200 Accounts a Leaf Scores Best")
plt.show()
```

The worst cell is depth 12 with 25 accounts a leaf (0.7308): deep and small memorises.

In the 400 column, from depth 6 down, the score barely moves (0.7630 to 0.7640): big leaves stop the tree on their own.

```python
# the best pair, refitted on all 18,000 training accounts, on the validation accounts
pre_tree = pre_search.best_estimator_
print("Depth:", pre_tree.get_depth(), " Leaves:", pre_tree.get_n_leaves())
model_performance_classification(pre_tree, X_valid, y_valid)
```

52 leaves and a validation ROC AUC of 0.7633, against the full tree's 0.6161 with 2,750 leaves.

## Post-Pruning: Grow, Then Cut Back

**Post-pruning** grows the full tree, then cuts back the branches that add least.
Cost-complexity pruning scores each smaller tree as its leaves' impurity plus alpha times its number of leaves, so a larger alpha keeps fewer leaves.

```python
# every alpha at which the full balanced tree loses a branch
path = full_bal.cost_complexity_pruning_path(X_train, y_train)
print("Leaves in the full balanced tree:", full_bal.get_n_leaves())
print("Alphas on the pruning path:", len(path.ccp_alphas))
```

# Feature Importance and Scaling

## Impurity and Permutation Importance

**Impurity importance**: each column's share of all the Gini decrease the tree's questions made, on the training rows (`.feature_importances_`).

**Permutation importance**: how much validation ROC AUC falls when one column is shuffled, averaged over 10 shuffles.

```python
# both measures for the pre-pruned tree, sorted by impurity importance
perm = permutation_importance(pre_tree, X_valid, y_valid, scoring="roc_auc", n_repeats=10, random_state=SEED)
importance = pd.DataFrame({"impurity": pre_tree.feature_importances_, "permutation": perm.importances_mean},
                          index=FEATURES).sort_values("impurity", ascending=False)
importance.head(8)
```

```python
# Day 6's lollipops, twice, in the same column order
fig, axes = plt.subplots(1, 2, figsize=(13, 7), sharey=True)
labels = ["Impurity importance (share of the Gini decrease)", "Permutation importance (ROC AUC lost)"]
for ax, measure, colour, label in zip(axes, ["impurity", "permutation"], ["C0", "C1"], labels):
    ax.hlines(importance.index, 0, importance[measure], color=colour)
    ax.plot(importance[measure], importance.index, "o", color=colour)
    ax.axvline(0, color="black", linewidth=1)
    ax.set_xlabel(label)
axes[0].invert_yaxis()
fig.suptitle("Both Measures Put PAY_0 First; PAY_2 Is 8th by Impurity and 2nd by Permutation")
plt.show()
```

`PAY_0` is first in both: 0.7115 of the impurity decrease, and shuffling it costs 0.1130 of ROC AUC.

`PAY_2` is 8th by impurity and 2nd by permutation: its questions make little of the Gini decrease, yet shuffling it costs the second most on unseen accounts.

3 columns are never asked about, and 4 permutation values sit just below zero, which is noise.

### What Importance Does Not Mean

Not a cause: a column the tree leans on goes with default in this file, nothing more.

Not a direction: importance says how much a column is used, not which way it pushes.

Shared among related columns: `PAY_0` and `PAY_2` move together (correlation 0.66 on the training accounts), so the tree can lean on one and starve the other.

Impurity importance comes from the training rows, so it can reward questions that memorise; permutation importance is measured on unseen rows.

## Scaling Does Not Change a Tree

```python
# the pre-pruned settings again, with every column scaled first inside a pipeline
scaled_tree = Pipeline([("scale", StandardScaler()),
                        ("tree", DecisionTreeClassifier(class_weight="balanced", max_depth=8, min_samples_leaf=200,
                                                        random_state=SEED))])
scaled_tree.fit(X_train, y_train)

# the same output for every validation account?
print((scaled_tree.predict_proba(X_valid) == pre_tree.predict_proba(X_valid)).all())
```

The same output for all 6,000 validation accounts.

A split compares a value with a threshold, and scaling keeps every value's order, so the tree asks the same questions in new units.

### Exercise 4: AI check

**The Search That Saw Too Much**

You asked an AI assistant for a function that picks a balanced tree's depth by cross-validation and returns the tree.
This came back.
It runs, and its validation ROC AUC looks good.
Test it, then find the one bug.

```python
# written for this course in the style of an AI assistant's answer: it runs, and it has one bug
def tuned_tree(X_train, y_train, X_valid, y_valid):
    # search depths 3 to 8 by five-fold cross-validation, then return the best tree
    X_all = pd.concat([X_train, X_valid])
    y_all = pd.concat([y_train, y_valid])
    search = GridSearchCV(DecisionTreeClassifier(class_weight="balanced", random_state=SEED),
                          {"max_depth": [3, 4, 5, 6, 7, 8]}, cv=folds, scoring="roc_auc")
    search.fit(X_all, y_all)
    return search.best_estimator_

ai_tree = tuned_tree(X_train, y_train, X_valid, y_valid)
print(round(roc_auc_score(y_valid, ai_tree.predict_proba(X_valid)[:, 1]), 4))
```

```python
# the test: the tree must be fitted on the 18,000 training accounts only (its root holds every account it saw)
assert ai_tree.tree_.n_node_samples[0] == len(X_train), "the tree saw more than the training accounts"
```

Q. Fix the function so the search fits on the training accounts only, and store the fixed tree's validation ROC AUC in `ex4_auc`.

```python
# the fix: the search fits on the training accounts only, so the validation accounts stay unseen
def tuned_tree(X_train, y_train, X_valid, y_valid):
    search = GridSearchCV(DecisionTreeClassifier(class_weight="balanced", random_state=SEED),
                          {"max_depth": [3, 4, 5, 6, 7, 8]}, cv=folds, scoring="roc_auc")
    search.fit(X_train, y_train)
    return search.best_estimator_

fixed_tree = tuned_tree(X_train, y_train, X_valid, y_valid)
assert fixed_tree.tree_.n_node_samples[0] == len(X_train)
ex4_auc = roc_auc_score(y_valid, fixed_tree.predict_proba(X_valid)[:, 1])
print(round(ex4_auc, 4), fixed_tree.get_depth())
```
