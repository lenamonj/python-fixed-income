### Exercise 1

Q. A position holds 1,000,000 face of a bond with a 4.5 percent coupon. What is the annual coupon income? Store it in **ex1_income**.

````python
# face times the coupon in percent, divided by 100
ex1_income = 1000000 * 4.5 / 100
print(ex1_income)
````

* 45,000 a year. The coupon is a percent, so it is divided by 100.

````python
check('Exercise 1', ex1_income, 45000.0)
````

### Exercise 2

Q. Use the `statistics` library to find the average of 99.50, 100.25, and 101.00, and store it in **ex2_average**.

````python
import statistics

# import the library first, then call its mean function on a list of the prices
ex2_average = statistics.mean([99.50, 100.25, 101.00])
print(ex2_average)
````

* 100.25. Without the `import` line, Python would stop with `NameError: name 'statistics' is not defined`: the library exists, but this notebook was never told to use it.

````python
check('Exercise 2', ex2_average, 100.25)
````

### Exercise 3

Q. Count the days from 15 January 2026 to 15 July 2026 with `date`, and store the number in **ex3_days**.

````python
from datetime import date

# subtracting two dates gives a length of time; .days reads it as a whole number of days
ex3_days = (date(2026, 7, 15) - date(2026, 1, 15)).days
print(ex3_days)
````

* 181 days on the calendar. Bond day-count conventions, which count some months as 30 days, come in Course 2.

````python
check('Exercise 3', ex3_days, 181)
````

### Exercise 5: AI check

Q. Read the assistant's code, run it, find the bug, and store the number of rows of the fixed table in **ex5_rows**.

````python
# written for this course in the style of an AI assistant's answer. It contains one bug.
import Pandas as pd

bonds = pd.DataFrame({"ticker": ["TRNW", "OSMC", "CLDB"], "price": [96.25, 100.00, 104.50]})
print(bonds)
````

* The code stops at its first line: `ModuleNotFoundError: No module named 'Pandas'`. The library is installed, but its name is `pandas`, all lower case. Python's names are case-sensitive, so `Pandas` is a different name, and no library has it.
* The fix is one letter.

````python
# the library's name is pandas, in lower case
import pandas as pd

bonds = pd.DataFrame({"ticker": ["TRNW", "OSMC", "CLDB"], "price": [96.25, 100.00, 104.50]})
ex5_rows = len(bonds)
print(bonds)
print("rows:", ex5_rows)
````

* Three rows. An error message that names the problem is the easy kind of bug: read its last line first.

````python
check('Exercise 5', ex5_rows, 3)
````
