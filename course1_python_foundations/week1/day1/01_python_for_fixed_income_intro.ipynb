## 1.2 How a Notebook Runs

Before going further, it helps to know how a notebook keeps track of things. Python remembers every variable from the moment its cell is run until the notebook is restarted. It pays no attention to where a cell sits on the page, only to which cells have been run, and in what order.

Q. `cost` was worked out when `face` was 1,000,000. Since then `face` has been changed to 1,500,000. Has `cost` changed too?

````python
# print the two variables as they stand now
print(f'face is {face:,} and cost is ${cost:,.2f}')
# to get the cost of the larger position, the calculation has to be run again
cost_now = face * price / 100
print(f'Worked out again, the cost is ${cost_now:,.2f}')
````

* `cost` still holds 985,000.00. A variable holds a value, not a formula. When `face` changed, nothing that had been calculated from it changed with it.
* Only running the calculation again, with the new `face`, gives 1,477,500.00.

**Excel side by side.** This is the biggest difference from a worksheet. In Excel, a cell holding `=A1*A2/100` recalculates by itself the moment A1 changes. Python never recalculates on its own. A line of code runs when its cell is run, and only then.

Three habits follow from this.

* **Run the cells from top to bottom.** Each cell relies on the cells above it. The number in square brackets beside a code cell shows the order in which the cells were run.
* **Running a cell twice does its work twice.** The cell `face = face + 500000` took `face` from 1,000,000 to 1,500,000. Run it a second time and `face` becomes 2,000,000, with no error and no warning.
* **When in doubt, start again.** *Restart* clears everything Python remembers, and *Run All* runs every cell in order from the top. In VS Code both are buttons in the bar above the notebook, and Colab has the same commands in its Runtime menu. A notebook that gives the same answers after Restart and Run All does not depend on the order you happened to click in.

Every notebook in this course is written to be run from top to bottom, and that is the way to use them.

## 1.4 Strings

Much of a desk's data is text: tickers, ratings, identifiers, descriptions. A string behaves like a list of characters, so indexing, slicing, and `len()` all work on it.

Every bond has a 9-character identifier called a CUSIP. The first 6 characters identify the issuer, the next 2 the issue, and the last is a check digit. The CUSIP below is invented.

Q. Store the CUSIP of QVNE and pull out its three parts.

````python
# an invented CUSIP for QVNE
cusip = '99012QAB6'
# len() counts the characters
print(len(cusip))
# the first 6 characters: start at the beginning and stop before index 6
print(cusip[:6])
# the next 2 characters: index 6 and index 7
print(cusip[6:8])
# the last character
print(cusip[-1])
````

* 9 characters, split into 99012Q, AB, and 6. A slice with nothing before the colon starts at the beginning. With nothing after the colon, it runs to the end.

**Excel side by side.** With the CUSIP in A1, `=LEN(A1)` returns 9, `=LEFT(A1,6)` returns 99012Q, `=MID(A1,7,2)` returns AB, and `=RIGHT(A1,1)` returns 6. `MID` counts from 1, so its 7 is Python's index 6.

Text typed by people arrives untidy. A string has methods that hand back a cleaned-up copy.

Q. A rating arrives as ' bbb ', in lower case with a space on each side. A face amount arrives as the text '1,500,000'. Clean both.

````python
rating_text = ' bbb '
# strip() removes the spaces at both ends, and upper() changes the letters to capitals
rating_clean = rating_text.strip().upper()
print(rating_clean)
# lower() goes the other way
print(rating_clean.lower())

face_text = '1,500,000'
# replace() swaps one piece of text for another. Here every comma is replaced with nothing
face_number = int(face_text.replace(',', ''))
print(face_number)
# the two original strings are unchanged. The square brackets show where each one starts and ends
print(f'[{rating_text}] [{face_text}]')
````

* `' bbb '` became `'BBB'`, and the text `'1,500,000'` became the number 1500000. `int()` cannot read a number with commas in it, so the commas have to go first.
* A string cannot be changed in place. Each method hands back a new string and leaves the original alone, so the result has to be stored in a variable.

**Excel side by side.**

| Python | Excel | Result |
| --- | --- | --- |
| `rating_text.strip()` | `=TRIM(A1)` | bbb |
| `rating_text.strip().upper()` | `=UPPER(TRIM(A1))` | BBB |
| `rating_clean.lower()` | `=LOWER(A1)` | bbb |
| `face_text.replace(',', '')` | `=SUBSTITUTE(A1,",","")` | 1500000, still text |

## 1.7 List Comprehensions

The desk wants a new list holding the annual coupon income of each position.

We already have **face_list** and **coupon_list**.

One way is to start with an empty list, loop over the positions, and add each result to the list.

````python
income_list = []

for face, coupon in zip(face_list, coupon_list):
    income = face * coupon / 100
    income_list.append(income)

print(income_list)
````

A **list comprehension** produces the same list in one line.

* It builds a new list from an existing sequence.
* It is shorter than a loop, and once you are used to it, easier to read.

````python
income_list = [face * coupon / 100 for face, coupon in zip(face_list, coupon_list)]
print(income_list)
````

The comprehension has three parts:

1. `[ ]`: the square brackets say we are building a list
2. `face * coupon / 100`: the calculation to perform for each item
3. `for face, coupon in zip(face_list, coupon_list)`: where the items come from

#### Positional arguments, keyword arguments, and defaults

So far, every argument has been matched by its position: the first value goes to the first name, and so on. These are **positional arguments**.

An argument can also be passed by name, written as `name=value` in the call. That is a **keyword argument**. We have already used one: `sep=' | '` in `print()`.

A default value is a separate idea. In the `def` line, an argument can be given a default, also written as `name=value`. If the caller leaves that argument out, the default is used.

Q. The desk wants the total coupon income of its positions. Usually it wants a full year, but sometimes only part of a year. Write a function that takes a list of annual incomes and a number of months, where the number of months has a default value of 12.

````python
# months has a default value of 12, so the caller may leave it out
def total_income(incomes, months=12):
    """
    Take a list of annual coupon incomes and the number of months,
    and return the total income for that period.
    """
    # add up the annual incomes
    total = sum(incomes)
    # scale the annual total to the period asked for
    return total * months / 12
````

````python
# months is left out, so the default of 12 is used
print(f'Income for a full year: ${total_income(income_list):,.2f}')
# months is passed by name, as a keyword argument
print(f'Income for six months: ${total_income(income_list, months=6):,.2f}')
# when every argument is passed by name, the order no longer matters
print(f'Income for three months: ${total_income(months=3, incomes=income_list):,.2f}')
````

* 163,750.00 for the year, 81,875.00 for six months, and 40,937.50 for three. One function covers every period, and the usual case needs no extra typing.

**Note**: when calling a function, keyword arguments go after the positional arguments.

````python
print(total_income(months=6, income_list))
````

* A `SyntaxError`. Python stops before running anything, because a positional argument came after a keyword argument.

#### Three things you will see in other people's code

You do not need to write the next three yet. You do need to recognize them, because libraries, colleagues, and AI assistants use them all the time.

**Reading note 1: lambda.** A `lambda` is a small function written in one line, with no name and no `return`. It is mostly used as a `key=`.

````python
# the same sort, with the key written as a lambda in place of price_of
by_price = sorted(holdings, key=lambda holding: holding['Price'], reverse=True)
print([holding['Ticker'] for holding in by_price])
````

* Read `lambda holding: holding['Price']` as: take a holding, hand back its price. It does exactly what `price_of` does.

**Reading note 2: a star in front of an argument.** One star lets a function accept any number of values. They arrive inside the function as a tuple. By convention the name is `args`, so you will usually see it written `*args`.

````python
def add_face(*positions):
    """
    Take the face held in any number of positions
    and return the total.
    """
    return sum(positions)


# four values, then six: the same function takes both
print(f'{add_face(1500000, 500000, 750000, 250000):,}')
print(f'{add_face(1500000, 500000, 750000, 250000, 300000, 200000):,}')
````

* 3,000,000 for four positions and 3,500,000 for six, with no change to the function.

**Reading note 3: two stars in front of an argument.** Two stars let a function accept any number of keyword arguments. They arrive inside the function as a dictionary. By convention the name is `kwargs`, so you will usually see it written `**kwargs`.

````python
def net_cash(income, **other_cash):
    """
    Take the coupon income for a period and any other named cash amounts,
    and return the net cash for the period.
    """
    # other_cash is a dictionary. Add up its values, whatever they are called
    return income + sum(other_cash.values())


# money in is positive, money out is negative
print(f'${net_cash(81875.0, maturing_bond=250000, custody_fees=-1500):,.2f}')
````

* 330,375.00: six months of coupon, plus a maturing bond, less custody fees. The function did not need to know the names `maturing_bond` and `custody_fees` in advance.
* When you meet `*args` or `**kwargs` in a function's documentation, read them as 'any number of values' and 'any number of named settings'.

A function can return more than one value.

Q. Write a function that returns the total face, the total cost, and the total annual coupon income of the portfolio.

````python
def portfolio_summary(faces, prices, coupons):
    """
    Take lists of face amounts, prices, and coupons
    and return the total face, total cost, and total annual coupon income.
    """
    total_face = 0
    total_cost = 0
    total_income = 0
    for face, price, coupon in zip(faces, prices, coupons):
        total_face = total_face + face
        total_cost = total_cost + face * price / 100
        total_income = total_income + face * coupon / 100
    # return three values at once
    return total_face, total_cost, total_income
````

````python
# three values come back together, as a tuple, so we store them in three variables
portfolio_face, portfolio_cost, portfolio_income = portfolio_summary(face_list, price_list, coupon_list)
print('Portfolio Summary')
print(f'Total Face: {portfolio_face:,}')
print(f'Total Cost: ${portfolio_cost:,.2f}')
print(f'Annual Coupon Income: ${portfolio_income:,.2f}')
````

* In a few lines, the desk has a reusable summary of any portfolio it can describe with three lists.
