## Part 1. Install Python

Python is the engine that runs the course's code. Day 0 explains what that means. Here you only install it.

### Step 1.1: Download the installer

1. Open your web browser.
2. Go to **python.org/downloads/windows**.
3. Under **Stable Releases**, find the newest **Python 3.13** line. On 25 October 2026 that was Python 3.13.17.
4. Under it, click **Windows installer (64-bit)**.
5. The file downloads to your **Downloads** folder. Its name starts with `python-3.13` and ends with `-amd64.exe`.

The page also lists newer versions, such as 3.14. This course is built and tested on **3.13**, so pick 3.13.

The Python Releases for Windows page, with Windows installer (64-bit) under Python 3.13.17 boxed in red

### Step 1.2: Run the installer

1. Open your **Downloads** folder.
2. Double-click the file you just downloaded.
3. If Windows asks whether to let the app make changes, click **Yes**.

The Python installer opens.

### Step 1.3: Tick the box at the bottom, then click Install Now

This is the most important step in the whole setup.

1. Look at the **bottom** of the window. Tick **Add python.exe to PATH**.
2. Then click **Install Now**.

**Why the box matters.** It lets you type `python` in any window and have Windows find it. If you skip it, `python` will not be found later.

The Python installer's first screen, with Install Now and the ticked Add python.exe to PATH box in red

*Under Install Now, your screen shows a folder with your user name in it. It is painted out here. Leave the other box, Use admin privileges, as it is.*

### Step 1.4: Wait, then close

1. A green bar shows the progress. It can take a few minutes.

   The Python installer's Setup Progress page, with the green bar part way across

2. When it says **Setup was successful**, click **Close**.

   The Python installer's Setup was successful page, with Close boxed in red

You may also see an option called **Disable path length limit**. You can skip it.

### Step 1.5: Check that Python works

1. Click the **Start** button (the Windows logo).
2. Type `PowerShell`.
3. Click **Windows PowerShell**. A window with a text prompt opens.
4. Type `python --version` and press **Enter**.

```powershell
python --version
```

You should see `Python 3.13` followed by a number.

A PowerShell window: python --version and py --version both print Python 3.13.8

*Your number may differ from the picture. Any Python 3.13 is right.*

**If it does not work:**

* **"python is not recognized"**, or the **Microsoft Store** opens: Python is not on the PATH yet. Close PowerShell, open it again, and retry.
* Still not working? Try `py --version` instead. If that works, type `py` wherever this course says `python`.
* Neither works? Run the installer again and make sure **Add python.exe to PATH** is ticked.

## Part 4. Make the Course Environment

Day 0 explains what an environment is. In short: a folder called `.venv` inside the course folder, holding one copy of Python and the libraries this course uses. You make it once.

All the steps in this part are typed in VS Code's **terminal**.

### Step 4.1: Open the terminal

1. In VS Code, click **View**, then **Terminal**. Or use the shortcut: hold down **Ctrl** and press **~**. The **~** key is at the top left, under **Esc**.
2. A panel opens at the bottom. Its prompt ends with the course folder's name and a `>`.

You type a line at the prompt and press **Enter** to run it.

### Step 4.2: Allow PowerShell to run the activate script (once per computer)

Windows blocks scripts in PowerShell by default. Step 4.3 needs one script, so allow scripts for your own user account. Type this line and press **Enter**:

```powershell
Set-ExecutionPolicy -Scope CurrentUser RemoteSigned
```

* If PowerShell asks you to confirm, type `Y` and press **Enter**.
* It prints nothing when it works. No news is good news.
* You only do this once on each computer.

### Step 4.3: Make the environment and switch it on

Type these two lines, pressing **Enter** after each:

```powershell
python -m venv .venv
.venv\Scripts\Activate.ps1
```

* The first line makes the `.venv` folder. It takes a few seconds and prints nothing.
* The second line switches the environment on. The prompt now starts with **(.venv)**.

The VS Code terminal after python -m venv .venv and the activate line, with the prompt starting (.venv)

*Your prompt shows your own user name where the picture says `you`.*

### Step 4.4: Install the course's libraries

With **(.venv)** showing, type this line and press **Enter**:

```powershell
python -m pip install -r course1_python_foundations\requirements.txt
```

* `pip` downloads each library on the course's list and installs it into `.venv`.
* Lines scroll past for a minute or two. Wait until the prompt comes back.
* The last lines start with **Successfully installed** and a list of names.
* A **[notice]** about a new pip release is harmless. You can ignore it.

The terminal at the end of the install, listing installed libraries, with the (.venv) prompt back
